```python
import numpy as np
from datetime import datetime, timedelta
```

```python
file = open("WhatsApp Chat with Nakshatra.txt", "r", encoding="utf-8")
lines = file.readlines()
file.close()
print(lines[:10])
print(len(lines))
```

```python
messages = []

for line in lines:
    line = line.strip()

    if line == "":
        continue

    if len(line) >= 8 and line[2] == "/" and line[5] == "/":
        messages.append(line)
    else:
        if len(messages) > 0:
            messages[-1] = messages[-1] + " " + line

print(messages[:10])
print(len(messages))
```

```python
chat = []

for message in messages:

    parts = message.split(" - ", 1)

    if len(parts) < 2:
        continue

    date_time = parts[0]
    remaining = parts[1]

    if ":" not in remaining:
        continue

    sender, text = remaining.split(":", 1)

    data = {
        "date_time": date_time,
        "sender": sender.strip(),
        "text": text.strip()
    }

    chat.append(data)

print(chat[:5])
print(len(chat))
```

```python
chat = []

system_messages = 0
media_messages = 0
deleted_messages = 0

for message in messages:

    parts = message.split(" - ", 1)

    if len(parts) < 2:
        continue

    date_time = parts[0]
    remaining = parts[1]

    if ":" not in remaining:
        system_messages += 1
        continue

    sender, text = remaining.split(":", 1)

    sender = sender.strip()
    text = text.strip()

    if text == "<Media omitted>":
        media_messages += 1

    if text == "This message was deleted":
        deleted_messages += 1

    data = {
        "date_time": date_time,
        "sender": sender,
        "text": text
    }

    chat.append(data)

print("Real Messages :", len(chat))
print("System Messages :", system_messages)
print("Media Messages :", media_messages)
print("Deleted Messages :", deleted_messages)
```

```python
total_messages = len(chat)

people = {}

for message in chat:

    name = message["sender"]

    if name in people:
        people[name] += 1
    else:
        people[name] = 1

print("Total Messages :", total_messages)
print("Total Participants :", len(people))

for name in people:
    print(name, ":", people[name])
```

```python
names = list(people.keys())

for i in range(len(names)):
    for j in range(i + 1, len(names)):
        if people[names[j]] > people[names[i]]:
            temp = names[i]
            names[i] = names[j]
            names[j] = temp

print("Participant Activity")
print("-" * 35)

for name in names:
    percent = (people[name] * 100) / total_messages
    print(f"{name:15} {people[name]:5} messages   {percent:.2f}%")
```

```python
dates = []

for message in chat:
    date = message["date_time"].split(",")[0]
    d = datetime.strptime(date, "%d/%m/%y")
    dates.append(d)
start_date = dates[0]
end_date = dates[0]

for d in dates:
    if d < start_date:
        start_date = d
    if d > end_date:
        end_date = d

total_days = (end_date - start_date).days + 1

print("Start Date :", start_date.strftime("%d/%m/%Y"))
print("End Date   :", end_date.strftime("%d/%m/%Y"))
print("Total Days :", total_days)
```

```python
day_count = {}

for message in chat:
    day = message["date_time"].split(",")[0]

    if day in day_count:
        day_count[day] += 1
    else:
        day_count[day] = 1

most_day = ""
most_messages = 0

for day in day_count:
    if day_count[day] > most_messages:
        most_messages = day_count[day]
        most_day = day

print("Most Active Day :", most_day)
print("Messages :", most_messages)
```

```python
hour_count = {}

for i in range(24):
    hour_count[i] = 0

for message in chat:

    date_time = message["date_time"]

    time = date_time.split(",")[1].strip()

    hour = int(time.split(":")[0])

    if "pm" in time.lower() and hour != 12:
        hour = hour + 12

    if "am" in time.lower() and hour == 12:
        hour = 0

    hour_count[hour] += 1

peak_hour = 0
peak_messages = 0

for hour in hour_count:

    if hour_count[hour] > peak_messages:
        peak_messages = hour_count[hour]
        peak_hour = hour

print("Peak Hour :", peak_hour)
print("Messages :", peak_messages)
```

```python
participants = []

for name in people:
    participants.append(name)

matrix = np.zeros((len(participants), 24), dtype=int)

for message in chat:
    name = message["sender"]
    time = message["date_time"].split(",")[1].strip()
    hour = int(time.split(":")[0])

    if "pm" in time.lower() and hour != 12:
        hour = hour + 12

    if "am" in time.lower() and hour == 12:
        hour = 0

    row = participants.index(name)
    matrix[row][hour] += 1

print(matrix)
```

```python
highest = matrix.max()

q1 = highest * 0.25
q2 = highest * 0.50
q3 = highest * 0.75

print("      ", end="")

for i in range(24):
    print(f"{i:>2}", end=" ")

print()

for i in range(len(participants)):

    print(f"{participants[i]:8}", end=" ")

    for j in range(24):

        value = matrix[i][j]

        if value == 0:
            ch = "."
        elif value <= q1:
            ch = "░"
        elif value <= q2:
            ch = "▒"
        elif value <= q3:
            ch = "▓"
        else:
            ch = "█"

        print(ch, end="  ")

    print()
```

```python
stop_words = [
    "i","me","my","we","our","you","your","he","she","it","they",
    "is","am","are","was","were","be","been","being",
    "the","a","an","and","or","to","of","in","on","at","for",
    "this","that","with","as","by","from","have","has","had",
    "will","would","can","could","should","do","does","did",
    "bhai","bro","ok","okay","hmm","haan","ha","yes","no",

    "ahe","aahe","mi","tu","tumhi","apan","pn","mg","tr","na",
    "ka","ki","la","cha","che","chi","tar","kay","ho","nahi",
    "bara","mag","ata","te","ti","to","mala","tula","tyala",
    "tyanna","apla","apli","aple"
]
```

```python
words = {}

for message in chat:

    text = message["text"].lower()

    text = text.replace(".", " ")
    text = text.replace(",", " ")
    text = text.replace("!", " ")
    text = text.replace("?", " ")
    text = text.replace(":", " ")
    text = text.replace(";", " ")
    text = text.replace("(", " ")
    text = text.replace(")", " ")
    text = text.replace('"', " ")
    text = text.replace("'", " ")

    parts = text.split()

    for word in parts:

        if word in stop_words:
            continue

        if len(word) <= 2:
            continue

        if word == "<media":
            continue

        if word == "omitted>":
            continue

        if word in words:
            words[word] += 1
        else:
            words[word] = 1
```

```python
top_words = []

temp = words.copy()

for i in range(10):

    max_word = ""
    max_count = 0

    for word in temp:

        if temp[word] > max_count:
            max_count = temp[word]
            max_word = word

    if max_word == "":
        break

    top_words.append((max_word, max_count))

    del temp[max_word]

print(top_words)
```

```python
print("Top 10 Words")
print("-" * 35)

for word, count in top_words:

    bar = "█" * count

    if len(bar) > 30:
        bar = bar[:30]

    print(f"{word:15} {bar} {count}")
```

```python
for message in chat:

    dt = message["date_time"]

    dt = dt.replace("\u202f", " ")
    dt = dt.replace("\u2009", " ")
    dt = dt.replace("\xa0", " ")

    message["time"] = datetime.strptime(dt, "%d/%m/%y, %I:%M %p")

response = {}

for name in people:
    response[name] = []

for i in range(1, len(chat)):

    current = chat[i]
    previous = chat[i - 1]

    if current["sender"] != previous["sender"]:

        gap = current["time"] - previous["time"]

        minutes = gap.total_seconds() / 60

        response[current["sender"]].append(minutes)

print("Average Response Time")
print("-" * 35)

for name in response:

    if len(response[name]) == 0:
        print(name, ": No Data")

    else:

        total = 0

        for value in response[name]:
            total += value

        average = total / len(response[name])

        print(f"{name:15} {average:.2f} minutes")
```

```python
all_dates = []

for message in chat:

    d = message["time"].date()

    if d not in all_dates:
        all_dates.append(d)

all_dates.sort()

print("Longest Silent Streak")
print("-" * 35)

for person in people:

    active_days = []

    for message in chat:

        if message["sender"] == person:

            d = message["time"].date()

            if d not in active_days:
                active_days.append(d)

    longest = 0
    current = 0

    for day in all_dates:

        if day in active_days:
            current = 0
        else:
            current += 1

            if current > longest:
                longest = current

    print(f"{person:15} {longest} days")
```

```python
emoji = {}

emoji_list = ["😂","🤣","❤️","❤","😭","😍","😊","😅","😎","🔥","👍","🥲","💀"]

for message in chat:

    text = message["text"]

    for e in emoji_list:

        if e in text:

            if e in emoji:
                emoji[e] += text.count(e)
            else:
                emoji[e] = text.count(e)

best = ""
count = 0

for e in emoji:

    if emoji[e] > count:
        count = emoji[e]
        best = e

print("Most Used Emoji :", best)
print("Count :", count)
```

```python
caring_words = ["okay","safe","eat","sleep","take","care","please","reminder","water","forget"]
laugh_words = ["lol","lmao","haha","hehe","rofl","lmfao"]
meme_words = ["meme","reel","instagram","insta"]

stats = {}

for person in people:

    stats[person] = {
        "messages":0,
        "words":0,
        "night":0,
        "questions":0,
        "caps":0,
        "laugh":0,
        "care":0,
        "meme":0
    }

for message in chat:

    person = message["sender"]
    text = message["text"]
    time = message["time"]

    stats[person]["messages"] += 1

    word_list = text.split()

    stats[person]["words"] += len(word_list)

    if time.hour >= 23 or time.hour <= 4:
        stats[person]["night"] += 1

    if text.endswith("?"):
        stats[person]["questions"] += 1

    if len(text) > 2 and text.isupper():
        stats[person]["caps"] += 1

    if text.count("!") >= 2:
        stats[person]["caps"] += 1

    for word in word_list:

        w = word.lower()

        if w in laugh_words:
            stats[person]["laugh"] += 1

        if w in caring_words:
            stats[person]["care"] += 1

        if w in meme_words:
            stats[person]["meme"] += 1

    stats[person]["meme"] += text.count("😂")
    stats[person]["meme"] += text.count("🤣")
    stats[person]["meme"] += text.count("😭")
    stats[person]["meme"] += text.count("😅")
    stats[person]["meme"] += text.count("💀")
```

```python
print("Personality Archetypes")
print("-" * 50)

personality = {}
used = []

highest_messages = max(people.values())
lowest_messages = min(people.values())

for person in people:

    if people[person] == highest_messages:
        personality[person] = "The Spammer"
        used.append(person)
        break

for person in people:

    if people[person] == lowest_messages and person not in used:
        personality[person] = "The Ghost"
        used.append(person)
        break

story_person = ""
story_score = -1

night_person = ""
night_score = -1

question_person = ""
question_score = -1

care_person = ""
care_score = -1

laugh_person = ""
laugh_score = -1

meme_person = ""
meme_score = -1

for person in people:

    if person in used:
        continue

    total = stats[person]["messages"]

    avg_words = stats[person]["words"] / total

    if avg_words > story_score:
        story_score = avg_words
        story_person = person

    if stats[person]["night"] > night_score:
        night_score = stats[person]["night"]
        night_person = person

    if stats[person]["questions"] > question_score:
        question_score = stats[person]["questions"]
        question_person = person

    if stats[person]["care"] > care_score:
        care_score = stats[person]["care"]
        care_person = person

    if stats[person]["laugh"] > laugh_score:
        laugh_score = stats[person]["laugh"]
        laugh_person = person

    if stats[person]["meme"] > meme_score:
        meme_score = stats[person]["meme"]
        meme_person = person

if story_person != "" and story_person not in used:
    personality[story_person] = "The Storyteller"
    used.append(story_person)

if night_person != "" and night_person not in used:
    personality[night_person] = "The Night Owl"
    used.append(night_person)

if question_person != "" and question_person not in used:
    personality[question_person] = "The Question Master"
    used.append(question_person)

if care_person != "" and care_person not in used:
    personality[care_person] = "The Caretaker"
    used.append(care_person)

if laugh_person != "" and laugh_person not in used:
    personality[laugh_person] = "The Comedian"
    used.append(laugh_person)

if meme_person != "" and meme_person not in used:
    personality[meme_person] = "The Meme Lord"
    used.append(meme_person)

for person in people:

    if person not in used:
        personality[person] = "The All Rounder"

print()

for person in people:
    print(person, ":", personality[person])
```
